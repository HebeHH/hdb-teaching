```python
resale_df = pd.read_csv("./data/ResaleflatpricesbasedonregistrationdatefromJan2017onwards.csv")
display(resale_df.head(10))
```

```python

town_items = set(resale_df["town"])
flat_type_items = set(resale_df["flat_type"])
flat_model_items = set(resale_df["flat_model"])
storey_range_items = set(resale_df["storey_range"])

print("Unique Towns:", town_items)
print("Unique Flat Types:", flat_type_items)
print("Unique Flat Models:", flat_model_items) 
print("Unique Storey Ranges:", storey_range_items)
```

```python
town_model = (
    resale_df
    .groupby(["town", "flat_model"])
    .size()
    .reset_index(name="count")
)

town_order = (
    resale_df["town"]
    .value_counts()
    .index
    .tolist()
)

display(town_model.head(10))
```

```python

fig = px.bar(
    town_model,
    x="town",
    y="count",
    color="flat_model",
    barmode="stack",
    category_orders={"town": town_order},
    title="HDB Resale Transactions by Town and Flat Model",
    labels={
        "town": "Town",
        "count": "Transactions",
        "flat_model": "Flat Model"
    },
    height=650
)

fig.update_layout(xaxis_tickangle=-45)
fig.show()
```

```python
fig = px.sunburst(
    resale_df,
    path=["town", "flat_type", "flat_model"],
    title="HDB Resale Distribution: Town → Flat Type → Model",
    height=850
)

fig.show()
```

```python
for column, label in [
    ("town", "Town"),
    ("flat_type", "Flat Type"),
    ("flat_model", "Flat Model")
]:
    counts = (
        resale_df[column]
        .value_counts()
        .reset_index()
    )

    counts.columns = [column, "count"]

    fig = px.bar(
        counts,
        x="count",
        y=column,
        orientation="h",
        title=f"HDB Resale Transactions by {label}",
        labels={
            column: label,
            "count": "Transactions"
        },
        text="count",
        height=max(450, len(counts) * 32)
    )

    fig.update_layout(
        yaxis={
            "categoryorder": "total ascending"
        },
        showlegend=False
    )

    fig.show()
```

```python
heatmap_df = pd.crosstab(
    resale_df["town"],
    resale_df["flat_type"]
)

fig = px.imshow(
    heatmap_df,
    text_auto=True,
    aspect="auto",
    color_continuous_scale="Viridis",
    title="HDB Resale Transactions: Town vs Flat Type",
    labels={
        "x": "Flat Type",
        "y": "Town",
        "color": "Transactions"
    },
    height=850
)

fig.show()
```
